## 1. Regrouper dans une liste

Une **liste** range plusieurs éléments dans un ordre précis, entre crochets `[ ]`, séparés par des virgules. Ici, chaque élément de la liste est un dictionnaire (un post).

```python
posts = [post_1, post_2, post_3, post_4, post_5, post_6, post_7, post_8, post_9, post_10]
print(posts)
```

**Accéder à un élément** : chaque élément a une **position** (un index), qui commence à **0**, pas à 1.

```python
print(posts[0])   # le premier élément de la liste : post_1
print(posts[0]["auteur"])   # sa valeur "auteur"
```

**Compter les éléments** : `len()` donne le nombre d'éléments d'une liste.

```python
print(len(posts))
```

## 2. Parcourir la liste avec for

Plutôt que d'écrire `posts[0]`, `posts[1]`, `posts[2]`... à la main, une **boucle `for`** répète un bloc de code pour chaque élément de la liste, automatiquement.

```python
for post in posts:
    print(post["auteur"])
```

À chaque tour de boucle, la variable `post` prend la valeur de l'élément suivant dans `posts`. Le code indenté (`print(...)`) s'exécute une fois par post — ici, 10 fois, sans qu'on ait eu à écrire `post_1`, `post_2`...

## 3. Combiner boucle et condition

On peut réutiliser une condition (séance 5) **à l'intérieur** d'une boucle, pour appliquer une règle à chaque post automatiquement.

```python
for post in posts:
    if post["likes"] > 100:
        print(post["auteur"], "- populaire")
```

## 4. Garder les résultats dans une nouvelle liste

Jusqu'ici, on a seulement **affiché** les posts qui correspondent à une règle. Pour les **garder** (les réutiliser ensuite), on les ajoute un par un à une liste vide, avec `.append()`.

```python
posts_populaires = []

for post in posts:
    if post["likes"] > 100:
        posts_populaires.append(post)

print(posts_populaires)
print(len(posts_populaires), "posts trouvés")
```

- `posts_populaires = []` crée une liste vide.
- `.append(post)` ajoute le post courant à la fin de cette liste, seulement s'il passe la condition.
