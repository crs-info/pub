| Opérateur | Signification |
|---|---|
| `>` | strictement supérieur |
| `<` | strictement inférieur |
| `>=` | supérieur ou égal |
| `<=` | inférieur ou égal |
| `==` | égal (attention : deux signes `=`, pas un seul) |
| `!=` | différent |

## 2. Décider avec if / else

Une comparaison donne juste `True` ou `False`. Pour que Python **fasse quelque chose** selon le résultat, on utilise `if` (si) et `else` (sinon).

Attention à la syntaxe :
- deux points `:` à la fin de la ligne `if ...` et `else :`
- le code à exécuter est **indenté** (décalé de 4 espaces)

```python
if post["likes"] > 50:
    print("à montrer en premier")
else:
    print("pas en premier")
```

## 3. Plusieurs cas avec elif

Si une règle a **plus de deux issues**, `elif` ("sinon si") permet d'ajouter des cas intermédiaires, testés dans l'ordre, du premier au dernier.

```python
if post["likes"] > 200:
    print("très populaire")
elif post["likes"] > 50:
    print("populaire")
else:
    print("peu populaire")
```

**Important** : Python s'arrête au premier cas vrai. L'ordre des conditions compte — c'est exactement la notion de **priorité** vue en séance 4.

## 5. Combiner plusieurs critères (and / or)

Certaines règles reposent sur **plusieurs critères à la fois** (ex. "compte favori ET plus de 50 likes"). On combine alors deux conditions avec `and` (et) ou `or` (ou).

- `and` → les deux conditions doivent être vraies
- `or` → au moins une des deux conditions suffit

```python
if post_1["auteur"] == "Léa" and post_1["likes"] > 50:
    print("compte favori et populaire")
else:
    print("ne correspond pas aux deux critères")
```

## 6. Et si votre règle est un classement ?

Certaines règles ne se répondent pas par oui ou non pour un post isolé — elles **comparent les posts entre eux**. Par exemple : *"les posts avec le plus de likes d'abord"*.

Ce type de règle s'appelle un **classement** : il n'a de sens que sur un groupe de posts, pas sur un post pris seul. `if` / `else` ne sait répondre que oui ou non pour **un seul post à la fois** — il ne peut pas encore faire ça.

**Pour aujourd'hui**, si une de vos règles est un classement, reformulez-la temporairement en remplaçant "le plus" par un nombre précis :

- *"Les posts avec le plus de likes d'abord"* → *"Les posts avec plus de 100 likes"*
- *"Les plus commentés d'abord"* → *"Les posts avec plus de 10 commentaires"*

Codez cette version temporaire avec `if` / `elif` / `else`, comme vos autres règles.


```python
# "les plus likés d'abord" devient temporairement :
if post["likes"] > 100:
    print("à montrer en premier")
else:
    print("pas en premier")
```
