### Comment organiser les données d'un post ?

Tout à l'heure, vous avez créé sur papier une **fiche** pour un post : une liste d'informations, chacune avec un nom et une valeur.

Par exemple :

| Information | Valeur |
|---|---|
| auteur | Marie |
| type | photo |
| likes | 42 |

En Python, cette structure a un nom : le **dictionnaire**. Il permet d'associer une **clé** (le nom de l'information) à une **valeur**.

## 1. Créer un dictionnaire

Voici comment on écrit la fiche du post ci-dessus en Python :

```python
post = {
    "auteur": "Marie",
    "type": "photo",
    "likes": 42,
}
```

- `post` est le nom qu'on donne à notre dictionnaire.
- Chaque ligne est une **clé** associée à une **valeur**.
- Les clés et les valeurs textuelles s'écrivent entre guillemets `" "`. Les nombres s'écrivent sans guillemets.

**Exécutez la cellule ci-dessous** (Shift + Entrée).

````python
post = {
    "auteur": "Marie",
    "type": "photo",
    "likes": 42,
}
````

## 2. Afficher un dictionnaire

Pour voir le contenu d'un dictionnaire, on utilise `print()`.

````python
print(post)
````

On peut aussi afficher **une seule valeur**, en donnant la clé entre crochets `[ ]` :

````python
print(post["auteur"])
````
