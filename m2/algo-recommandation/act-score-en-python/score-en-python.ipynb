## 4. Classer les posts

Chaque post a maintenant un score. Il reste à les **classer** : du plus grand score au plus petit.

La cellule ci-dessous contient un outil de classement tout prêt. **Exécutez-la**, sans la modifier.

```python
# ⚠️ À EXÉCUTER, NE PAS MODIFIER
# Ce code classe les posts du plus grand score au plus petit.
# Vous n'avez pas besoin de le comprendre aujourd'hui : nous l'aborderons à la prochaine séance.

def classer(liste):
    reste = liste.copy()
    classement = []
    while len(reste) > 0:
        meilleur = reste[0]
        for post in reste:
            if post["score"] > meilleur["score"]:
                meilleur = post
        classement.append(meilleur)
        reste.remove(meilleur)
    return classement

def afficher_top3(liste):
    classement = classer(liste)
    for i in range(3):
        post = classement[i]
        print(i + 1, "-", post["auteur"], "(post", post["id"], ") :", post["score"], "points")

print("Outil de classement prêt.")
```

Pour afficher les 3 premiers posts du fil, il suffit d'écrire `afficher_top3(posts)`. **Exécutez la cellule.**

```python
afficher_top3(posts)
```

## 5. Changer les poids

Dans la formule, le nombre `2` devant les commentaires est un **poids** : il dit combien un critère compte.

On peut aussi **pénaliser l'ancienneté** : enlever des points aux posts anciens. Plus le poids de l'ancienneté est grand, plus un vieux post perd de points.

> score = likes + commentaires × poids_commentaires − anciennete_h × poids_anciennete

Dans la cellule ci-dessous, les poids sont des **variables**. **Changez seulement les deux nombres**, puis exécutez la cellule.

```python
poids_commentaires = 2
poids_anciennete = 0

for post in posts:
    post["score"] = post["likes"] + post["commentaires"] * poids_commentaires - post["anciennete_h"] * poids_anciennete

afficher_top3(posts)
```
